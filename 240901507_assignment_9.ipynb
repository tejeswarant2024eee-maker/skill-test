### 4. Euclidean Distance for Two Points in 2D Space

```python
import math

def euclidean_distance(point1, point2):
    """
    Calculates the Euclidean distance between two points in 2D space.

    Args:
        point1 (tuple): A tuple representing the first point (x1, y1).
        point2 (tuple): A tuple representing the second point (x2, y2).

    Returns:
        float: The Euclidean distance between the two points.
    """
    x1, y1 = point1
    x2, y2 = point2

    distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
    return distance

# Example Usage for Euclidean Distance
point_a = (1, 2)
point_b = (4, 6)
dist = euclidean_distance(point_a, point_b)
print(f"The Euclidean distance between {point_a} and {point_b} is: {dist}")

point_c = (0, 0)
point_d = (3, 4)
dist2 = euclidean_distance(point_c, point_d)
print(f"The Euclidean distance between {point_c} and {point_d} is: {dist2}")
```

### 4. Dot Product for Two Matrices

```python
def dot_product_matrix(matrix1, matrix2):
    """
    Calculates the dot product (matrix multiplication) of two matrices.

    Args:
        matrix1 (list of lists): The first matrix.
        matrix2 (list of lists): The second matrix.

    Returns:
        list of lists: The resulting matrix from the dot product, or an error message
                       if matrices are incompatible.
    """
    rows1 = len(matrix1)
    cols1 = len(matrix1[0])
    rows2 = len(matrix2)
    cols2 = len(matrix2[0])

    # Check for compatibility
    if cols1 != rows2:
        return "Error: Number of columns in the first matrix must be equal to the number of rows in the second matrix."

    # Initialize the result matrix with zeros
    result_matrix = [[0 for _ in range(cols2)] for _ in range(rows1)]

    # Perform matrix multiplication
    for i in range(rows1):
        for j in range(cols2):
            for k in range(cols1):
                result_matrix[i][j] += matrix1[i][k] * matrix2[k][j]
    return result_matrix

# Example Usage for Dot Product of Matrices
matrix_a = [
    [1, 2, 3],
    [4, 5, 6]
]

matrix_b = [
    [7, 8],
    [9, 1],
    [2, 3]
]

product = dot_product_matrix(matrix_a, matrix_b)
print("Matrix A:")
for row in matrix_a: print(row)
print("\nMatrix B:")
for row in matrix_b: print(row)
print("\nDot Product of Matrix A and Matrix B:")
if isinstance(product, str):
    print(product)
else:
    for row in product: print(row)


# Example with incompatible matrices
matrix_c = [[1, 2], [3, 4]]
matrix_d = [[1, 2, 3]]

incompatible_product = dot_product_matrix(matrix_c, matrix_d)
print("\nMatrix C:")
for row in matrix_c: print(row)
print("\nMatrix D:")
for row in matrix_d: print(row)
print("\nDot Product of Matrix C and Matrix D (incompatible):")
print(incompatible_product)
```
