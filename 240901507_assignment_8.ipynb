3. Write a Python script for Flight Data Analysis:

Problem Statement:

You have a CSV file named 'flight_data.csv' containing flight information such as origin,

destination, and departure time.

Perform the following tasks:

a. Display the summary statistics for departure delays,

b. Calculate the average delay for each airline,

c. Identify the most common departure and arrival destinations,

d. Visualize the distribution of departure delays using a box plot.

```python
import pandas as pd
import io
import matplotlib.pyplot as plt
import seaborn as sns

# Create sample flight data as a string
csv_data = """Flight_ID,Origin,Destination,Departure_Time,Departure_Delay,Airline
F001,JFK,LAX,08:00,15,Delta
F002,LAX,SFO,09:30,-5,American
F003,JFK,MIA,10:00,30,United
F004,SFO,JFK,11:15,0,Delta
F005,MIA,LAX,12:00,20,American
F006,LAX,JFK,13:45,-10,United
F007,JFK,LAX,14:00,60,Delta
F008,SFO,MIA,15:30,5,American
F009,MIA,JFK,16:00,45,United
F010,LAX,SFO,17:00,0,Delta
F011,JFK,LAX,18:00,10,American
F012,SFO,JFK,19:00,-2,United
"""

# Load the data into a pandas DataFrame
data_io = io.StringIO(csv_data)
df_flights = pd.read_csv(data_io)

print("Flight Data Head:")
display(df_flights.head())
```

### a. Display the summary statistics for departure delays

```python
delay_summary = df_flights['Departure_Delay'].describe()
print("Summary Statistics for Departure Delays:")
display(delay_summary)
```

### b. Calculate the average delay for each airline

```python
average_delay_by_airline = df_flights.groupby('Airline')['Departure_Delay'].mean().reset_index()
print("Average Delay for Each Airline:")
display(average_delay_by_airline)
```

### c. Identify the most common departure and arrival destinations

```python
most_common_origin = df_flights['Origin'].value_counts().idxmax()
most_common_destination = df_flights['Destination'].value_counts().idxmax()

print(f"Most Common Departure Origin: {most_common_origin}")
print(f"Most Common Arrival Destination: {most_common_destination}")
```

### d. Visualize the distribution of departure delays using a box plot

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x=df_flights['Departure_Delay'], color='skyblue')
plt.title('Distribution of Departure Delays')
plt.xlabel('Departure Delay (minutes)')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```
